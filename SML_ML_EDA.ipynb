```
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.metrics import r2_score
```

## 1. Basic Histogram
Generate 250 random values and draw a histogram with 5 bars.

```
x = np.random.uniform(0.0, 5.0, 250)

plt.hist(x, bins=5)
plt.title("Basic Histogram (250 random values, 5 bars)")
plt.show()
```

## 2. Big Data Distribution — Normal Distribution Histogram
Generate 100,000 values from a normal distribution (mean=5.0, std=1.0) and draw a histogram with 100 bars.

```
x = np.random.normal(5.0, 1.0, 100000)

plt.hist(x, bins=100)
plt.title("Normal Distribution Histogram (100,000 values, 100 bars)")
plt.show()
```

## 3. Scatter Plot — Basic Example
x = age of each car, y = speed of each car.

```
x = [5,7,8,7,2,17,2,9,4,11,12,9,6]
y = [99,86,87,88,111,86,103,87,94,78,77,85,86]

plt.scatter(x, y)
plt.title("Scatter Plot: Age vs Speed")
plt.show()
```

## 4. Scatter Plot — Random Data Distributions
Two arrays of 1000 random values each, from normal distributions with different means/std devs.

```
x = np.random.normal(5.0, 1.0, 1000)
y = np.random.normal(10.0, 2.0, 1000)

plt.scatter(x, y)
plt.title("Scatter Plot of Two Random Normal Distributions")
plt.show()
```

## 5. Linear Regression
Find the relationship between age (x) and speed (y) of 13 cars, and draw the regression line.

```
x = [5,7,8,7,2,17,2,9,4,11,12,9,6]
y = [99,86,87,88,111,86,103,87,94,78,77,85,86]

slope, intercept, r, p, std_err = stats.linregress(x, y)

def myfunc(x):
    return slope * x + intercept

mymodel = list(map(myfunc, x))

plt.scatter(x, y)
plt.plot(x, mymodel)
plt.title("Linear Regression: Age vs Speed")
plt.show()

print("r (correlation coefficient):", r)
```

### Predict a future value
Use the model to predict the speed of a car that is 10 years old.

```
speed = myfunc(10)
print("Predicted speed for a 10-year-old car:", speed)
```

## 6. Linear Regression — Bad Fit Example
A data set where linear regression is not a good fit.

```
x = [89,43,36,36,95,10,66,34,38,20,26,29,48,64,6,5,36,66,72,40]
y = [21,46,3,35,67,95,53,72,58,10,26,34,90,33,38,20,56,2,47,15]

slope, intercept, r, p, std_err = stats.linregress(x, y)

def myfunc(x):
    return slope * x + intercept

mymodel = list(map(myfunc, x))

plt.scatter(x, y)
plt.plot(x, mymodel)
plt.title("Linear Regression: Bad Fit Example")
plt.show()

print("r (correlation coefficient):", r)
print("A value close to 0 indicates a very bad relationship / bad fit.")
```

## 7. Polynomial Regression
Fit a polynomial (degree 3) model to a curved data set.

```
x = [1,2,3,5,6,7,8,9,10,12,13,14,15,16,18,19,21,22]
y = [100,90,80,60,60,55,60,65,70,70,75,76,78,79,90,99,99,100]

mymodel = np.poly1d(np.polyfit(x, y, 3))

myline = np.linspace(1, 22, 100)

plt.scatter(x, y)
plt.plot(myline, mymodel(myline))
plt.title("Polynomial Regression (degree 3)")
plt.show()

print("R-squared:", r2_score(y, mymodel(x)))
```

### Predict a future value
Use the polynomial model to predict a value at x = 17.

```
speed = mymodel(17)
print("Predicted value at x = 17:", speed)
```

## 8. Polynomial Regression — Bad Fit Example
A data set where polynomial regression is not a good fit.

```
x = [89,43,36,36,95,10,66,34,38,20,26,29,48,64,6,5,36,66,72,40]
y = [21,46,3,35,67,95,53,72,58,10,26,34,90,33,38,20,56,2,47,15]

mymodel = np.poly1d(np.polyfit(x, y, 3))

myline = np.linspace(2, 95, 100)

plt.scatter(x, y)
plt.plot(myline, mymodel(myline))
plt.title("Polynomial Regression: Bad Fit Example")
plt.show()

print("R-squared:", r2_score(y, mymodel(x)))
print("A value close to 0 indicates a very bad relationship / bad fit.")
```
